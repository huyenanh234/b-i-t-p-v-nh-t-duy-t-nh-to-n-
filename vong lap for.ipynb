```python
#kiem tra 1 so co phai so nguyen to khong

n = int(input())
if n<2 :
    print(f"{n} khong la so nguyen to")
else :
    t = 2
    check = True
    while t*t < n :
        if n % t== 0:
            check = False 
        t += 1
    if check :
        print(f"{n} la so nguyen to")
    else :
        print(f"{n} khong la so nguyen to ")
```
