```python
#Bài 1: Nhập vào 2 cạnh của một hình chữ nhật. In ra màn hình diệntíchvàchu vi của nó

a, b = map(int, input().split())
chu_vi = (a+b)*2
dien_tich = a*b
print(chu_vi, dien_tich)
```

```python
#Bài 2: Nhập vào bán kính của hình tròn. In ra màn hình diện tích và chu vi của nó. 
import math 
R = float(input())
match R :
    case _ if R<0:
        print(f"{R} khong phai ban kinh hinh tron")
    case _:
        chu_vi_ht = 2*math.pi*R
        dientichhinhtron = math.pi * (R**2)
        print(chu_vi_ht, dientichhinhtron)
```

```python
# Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là 
# độ dài ba cạnh hay không, nếu  không thì in  ra màn  hình  'Khong la ba  canh 
# cua mot tam  giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, 
# Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn 
# hình.

a, b, c = map(float, input().split())
A = (a, b, c)
match A :
    case _ if a>0 and b>0 and c>0 and a+b>c and b+c>a and a+c>b :
        chuvitamgiac = a + b + c
        p = chuvitamgiac/2
        dientichtamgiac = (p*(p-a)*(p-b)*(p-c))**0.5
        print(chuvitamgiac, dientichtamgiac)
        if a == b == c :
            print("A la tam giac deu")
        elif a==b or a==c or b==c :
            print("A la tam giac can")
        elif a**2 + b**2 == c**2 or b**2 + c**2 == a**2 or a**2 + c**2 == b**2:
            print("A la tam giac vuong")
        else :
            print("A la tam giac thuong")
    case _:
        print(f"{a}, {b}, {c} khong la ba canh cua tam giac")
        
```

```python
# Bài 4: Viết chương trình giải phương trình bậc 1 (ax + b = 0)  

a, b = map(float, input().split())
if a == 0 :
    if b == 0 :
        print("phuong trinh bac nhat vo so nghiem")
    else :
        print("phuong trinh vo nghiem")

else :
    print("phuong trinh co nghiem duy nhat x = -b/2a")
```

```python
# Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0) 

a, b, c = map(float, input().split())
if a != 0 :
    delta = b**2 - 4*a*c 
    if delta > 0 :
        print(f"phuong trinh {a}x**2+{b}x+{c}=0 co 2 nghiem x1 =(({-b}+{delta**0.5})/{2*a}) va x2=(({-b}-{delta**0.5})/{2*a})")
    elif delta < 0 :
        print(f"phuong trinh {a}x**2 +{b}x +{c}=0 vo nghiem")
    else :
        print(f"phuong trinh {a}x**2 +{b}x +{c}=0 co nghiem kep x = {-b}/{2*a}")
else :
    print("Phuong trinh khong phai dang bac 2")
```
