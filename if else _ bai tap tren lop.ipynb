```python
# kiem tra ngay thang nam xem co hop le khong

ngay, thang, nam = map(int, input().split())
# ngày từ 1 đến 31, tháng từ 1 đến 12, năm lớn hơn 1900
# kiểm tra các tháng đặc biệt
# kiểm tra năm nhuận xem tháng 2 có 28 hay 29 ngày

if ngay in range(1, 32) and thang in range(1, 13) and nam>=1900 :
    match thang :
        case 1|3|5|7|8|10|12 :
            print(f"{ngay}/{thang}/{nam} ton tai")
        case 6|4|9|11 :
            if ngay in range(1, 31):
                print(f"{ngay}/{thang}/{nam} ton tai")
            else :
                print(f"{ngay}/{thang}/{nam} khong hop le")
        case 2 :
            if (nam%4==0 and nam%10 ) or nam%100==0 :
                if ngay in range(1, 30):
                    print(f"{ngay}/{thang}/{nam} ton tai")
                else:
                    print(f"{ngay}/{thang}/{nam} khong hop le")
            else :
                if ngay in range(1, 29):
                    print(f"nam {nam} la nam thuong")
                else :
                    print(f"nam {nam} khong co ngay 29/2")

else :
    print(f"{ngay}/{thang}/{nam} khong ton tai")
```

```python
#[Rounding] 
# Nhập vào một số thực: in ra kết quả làm tròn lên, tròn xuống, làm tròn tới số 
# nguyên gần nhất (không được sử dụng hàm sẵn có của pyton). Các kết quả 
# cách nhau một dấu cách
n = float(input())
x = int(n)
if n>x :
    a = x + 1 
else :
    a = x 

if n<x :
    b = x - 1 
else :
    b = x 

if abs(n-x)>=0.5:
    if n >= 0 :
        c = x + 1 
    else :
        c = x - 1 
else :
    c = x 

print(a, b, c)
```
